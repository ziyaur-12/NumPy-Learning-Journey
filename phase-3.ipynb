## Advance Operation with Business Example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structures :[restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 1000, 1200, 1500, 1800], # paradise birani
    [2, 8000, 9000, 1000, 3000], #Burger king
    [3, 1500, 1600, 1700, 2000], #Taco Bell
    [4, 2000, 2200, 2500, 3000], #KFC
    [5, 5000, 6000, 7000, 8000] #McDonald's
])

print("Sales Data:\n", sales_data)
print("Shape of the sales data:", sales_data.shape)
```

```python
# sample data of first 3 restaurants
sample_data = sales_data[:3]
print("Sample Data (First 3 Restaurants):\n", sample_data)

# Get the sales data for a specific restaurant
restaurant_1_data = sales_data[0, 1:]
print("Sales Data for Restaurant 1:\n", restaurant_1_data)

restaurant_2_data = sales_data[1, 1:]
print("Sales Data for Restaurant 2:\n", restaurant_2_data)

restaurant_3_data = sales_data[2, 1:]
print("Sales Data for Restaurant 3:\n", restaurant_3_data)

print("sales data for 3rd restaurant in 2023:", sales_data[2, 3])
print("sales data for restaurant 4", sales_data[3:4]) # This will print the sales data for the 4th restaurant from 2021 to 2024
```

## Yearly Total|

```python
# total sales per year
print(np.sum(sales_data, axis=0)) # This will print the total sales for each year from 2021 to 2024 

#or 
yearly_sales = np.sum(sales_data[:, 1:], axis=0)
print("Total Sales per Year (2021-2024):", yearly_sales)    
```

```python
# minimum sales per year
min_sales_per_year = np.min(sales_data[:, 1:], axis=0)
print("Minimum Sales per Year (2021-2024):", min_sales_per_year)

# maximum sales per year
max_sales_per_year = np.max(sales_data[:, 1:], axis=0)
print("Maximum Sales per Year (2021-2024):", max_sales_per_year)
```

```python
# Average sales per year
average_sales_per_year = np.mean(sales_data[:, 1:], axis=0)
print("Average Sales per Year (2021-2024):", average_sales_per_year)
```

```python
# Cumulative sales per restaurant
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print("Cumulative Sales per Restaurant:\n", cumsum)
```

```python
# Graphical representation of sales data using matplotlib
plt.figure(figsize=(10, 6))
# Plotting sales data for each restaurant
plt.plot(np.mean(cumsum, axis = 0), label='Average Sales', marker='o')
plt.title('Sales Data for Restaurants (2021-2024)')
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

## vector

```python
# Addition
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])
result = vector1 + vector2
print("Result of vector addition:", result)
```

```python
# Multiplication
result_multiplication = vector1 * vector2
print("Result of vector multiplication:", result_multiplication)
```

```python
# Dot Product
dot_product = np.dot(vector1, vector2)
print("Dot Product of vector1 and vector2:", dot_product)
```

```python
# Angle between two vectors
angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))   
print("Angle between vector1 and vector2 (in radians):", angle)
print("Angle between vector1 and vector2 (in degrees):", np.degrees(angle))
```

## Vectorized Operation

```python
restaurant_types = np.array(['Paradise Biryani', 'Burger King', 'Taco Bell', 'KFC', "McDonald's"])
vectorized_uppercase = np.vectorize(str.upper)
print("Restaurant Types in Uppercase:\n", vectorized_uppercase(restaurant_types))
```

## Broadcast

```python
monthly_avg = sales_data[:, 1:] / 12
print("Monthly Average Sales per Year (2021-2024):", monthly_avg)
```

```python
#monthly average sales per restaurant
monthly_avg_per_restaurant = np.mean(monthly_avg, axis=1)
print("Monthly Average Sales per Restaurant:", monthly_avg_per_restaurant)
```
