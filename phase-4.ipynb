```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1, 2, 3], [4, 5, 6]])
array2 = np.random.rand(3,3)
array3 = np.zeros((4, 4))
```

## Saving the arrays

```python
np.save('array1.npy', array1)
np.save('array2.npy', array2)
np.save('array3.npy', array3)
```

## loading the saved arrays

```python
loaded_array1 = np.load('array1.npy')
print("\n print the saved array1:\n", loaded_array1)

loaded_array2 = np.load('array2.npy')
print("\n print the saved array2:\n", loaded_array2)

loaded_array3 = np.load('array3.npy')
print("\n print the saved array3:\n", loaded_array3)
```

### Loading the .npy Logo

```python
try:
    logo = np.load('numpy-logo.npy')

    plt.figure(figsize=(10, 5))

    plt.subplot(121)
    plt.title("NumPy Logo")
    plt.grid(False)
    plt.imshow(logo)

except FileNotFoundError:
    print("Logo file not found")
```

## Dark color the logo

```python
dark_logo = 1 - logo
plt.subplot(122)
plt.title("Dark NumPy Logo")
plt.grid(False)
plt.imshow(dark_logo)
```
